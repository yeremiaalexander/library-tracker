## 0. Setup & Engine Connection

```python
import os
import pandas as pd
from sqlalchemy import create_engine, text

schema = "tester_lib"
host = "127.0.0.1"
user = "root"
password = os.getenv("MYSQL_PASSWORD")
port = 3306

connection_string = f"mysql+pymysql://{user}:{password}@{host}:{port}/{schema}"
engine = create_engine(connection_string)
```

```python
def read_books(available_only=False):
    books = pd.read_sql("books", con=engine)
    loans= pd.read_sql("loans", con=engine)

    if available_only == True:
        books = pd.merge(books, loans, on='isbn', how='left').query('friend_id.isna()')[books.columns]

    return books
```

```python
def create_friend(name,max_loans=5,notes=None):
    df = pd.DataFrame([[name,max_loans,notes]], columns=['name','max_loans','notes'])
    df.to_sql('friends', if_exists='append', con=engine, index=False) #index if you dont want to keep panda index
    return f"{name} is added to the list"
```

```python
# create a loan
def create_loan(friend_name, isbn, loan_date=None, next_contact=None, notes=None):
    friends_df = read_friends()
    books_df = read_books()
    loans_df = read_loans()

    friend_rows = friends_df[friends_df["name"] == friend_name]
    book_rows = books_df[books_df["isbn"] == isbn]

    if friend_rows.empty:
        raise ValueError(f"Friend not found: {friend_name}")
    if book_rows.empty:
        raise ValueError(f"Book not found: {isbn}")

    friend_id = int(friend_rows.iloc[0]["friend_id"])
    if not loans_df[(loans_df["isbn"] == isbn) & (loans_df["friend_id"] == friend_id)].empty:
        return f"Loan already exists for {friend_name} and {isbn}"

    loan = pd.DataFrame(
        [[isbn, friend_id, loan_date, next_contact, notes]],
        columns=["isbn", "friend_id", "loan_date", "next_contact", "notes"]
    )
    loan.to_sql("loans", con=engine, if_exists="append", index=False)

    return f"{friend_name} has borrowed {book_rows.iloc[0]['title']}"
```
